```python
import numpy as np
import pandas as pd

data = np.array([
    ['Male'],
    ['Female'],
    [np.nan],
    ['Male'],
    ['Male'],
    [np.nan],
    ['Female'],
    ['Male']
], dtype=object)
c1=0
c2=0
for i in data:
    if i=='Male':
        c1+=1
    else:
        c2+=1
if c1 > c2:
    data[pd.isna(data)] = 'Male'
else:
    data[pd.isna(data)] = 'Female'
data
```

```python
import numpy as np
a = np.array([
    [10, np.nan, 30],
    [20, 25, np.nan],
    [np.nan, 35, 40],
    [50, 55, 60]
])
for i in range(0,a.shape[0]):
    for j in range(a.shape[1]):
        if np.isnan(a[i,j]):
            a[i,j] = a[i,j-1]
print("forward fill:-\n",a)
```

```python
import numpy as np
a = np.array([
    [10, np.nan, 30],
    [20, 25, np.nan],
    [np.nan, 35, 40],
    [50, 55, 60]
])
for i in range(0,a.shape[0]):
    for j in range(a.shape[1]-1,-1,-1):
        if np.isnan(a[i,j]):
            if j==2:
                a[i,j] = a[i,0]
            else:
                a[i,j] = a[i,j+1]
print("backward fill:-\n",a)
```

```python
import numpy as np
marks = np.array([
    [85, 78, 92],
    [45, 55, 60],
    [95, 88, 91],
    [35, 40, 42],
    [72, 68, 75],
    [90, 92, 94],
    [50, 48, 52],
    [80, 85, 88],
    [65, 70, 72],
    [99, 100, 98]
])
total = np.array(np.sum(marks,axis=1))
print("top 5 students:-\n",marks[total.argsort()[::-1]][:5])
print("subject wise average:-",np.mean(marks,axis=0))
print("student below passing marks:-\n",marks[(marks<50).any(1)])
```
