```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
sales = [10, 20, 35, 25, 40]
profit = [5, 15, 20, 18, 30]

max_sales = max(sales)
max_index = sales.index(max_sales)

plt.figure(figsize=(8,5))
plt.plot(x,sales,color = 'blue',marker = 'o',label = 'sales')
plt.plot(x,profit,color = 'red',marker = 'o',label = 'profit')
plt.scatter(x[max_index],max_sales,color = 'black',label = 'max sales',s=150)
plt.legend()
plt.savefig('sales analysis.png',dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

subjects = ["Math", "Science", "English", "History", "Computer"]
marks = [85, 92, 78, 88, 95]

plt.figure(figsize=(8,5))
colors = []
high = max(marks)
low = min(marks)
for i in marks:
    if i==high:
        colors.append('green')
    elif i==low:
        colors.append('red')
    else:
        colors.append('black')
plt.barh(subjects,marks,color = colors)
plt.xlabel('subjects')
plt.ylabel('marks')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x = np.random.randint(1,50,20)
y = np.random.randint(1,25,20)
plt.figure(figsize = (8,5))
plt.scatter(x,y,marker = '^',color = 'green')
plt.text(x[np.argmax(x)],y[1],'max value at x',ha='center')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
